# Task

Write a program to remove duplicate elements from a list

```python
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune", "Earth", "Mercury", "Mars", "Mars"]

removed_duplicates = list(set(planets))
removed_duplicates # Order is not preserved
```

```python
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune", "Earth", "Mercury", "Mars", "Mars"]

removed_duplicates = []
for planet in planets:
  if (planet not in removed_duplicates):
    removed_duplicates.append(planet)

removed_duplicates
```
