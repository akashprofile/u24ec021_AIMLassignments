# Task

Write a function that returns the square of a number.

```python
# Function definition
def square(x):
  return x ** 2
```

```python
import random

random.seed(3)

l1 = []
for i in range(0, 5, 1):
  l1.append(random.uniform(0, 99))

l2 = []
for i in range(0, 5, 1):
  l2.append(random.randint(0, 99))

print(l1)
print(l2)
```

```python
for value in l1:
  print(f"{value:.4f} --> {square(value):.4f}")
```

```python
for value in l2:
  print(f"{value:2d} ---> {square(value):4d}")
```
