# Task

Write a program to find the second largest number in a list.

```python
import random

random.seed(4)

l = []
for _ in range(0, 10, 1):
  l.append(random.randint(0, 99))

l
```

```python
l = sorted(l)
```

```python
m = l[::-1] # Start at the end (default when step is negative)
m
```

```python
print(f"Second largest number is {m[1]}")
```
