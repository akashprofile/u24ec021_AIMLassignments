# Task

Create a list of 20 numbers. Find the maximum, minimum, sum, and average.

```python
import random

random.seed(1)

l = []
for i in range(0, 20, 1):
  l.append(random.randint(0, 99))

l
```

```python
max(l) # maximum value from list
```

```python
min(l) # minimum value from list
```

```python
sum = 0
for value in l:
  sum += value
sum
```

```python
average = sum / len(l)
average
```
