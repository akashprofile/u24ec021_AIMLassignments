# Task

Print all even and odd numbers from a given list.

```python
import random

random.seed(2)

l = []
for i in range(0, 20, 1):
  l.append(random.randint(0, 99))

l
```

```python
# Printing all evens
for value in l:
  if (value % 2 == 0):
    print(value, end=", ")
```

```python
# Printing all odds
for value in l:
  if (value % 2 != 0):
    print(value, end=", ")
```
