# Task

Create a dictionary containing information and print all keys and values.

```python
info = {
  "firstName": "Akash",
  "lastName": "Kumar",
  "branch": "ECE",
  "cgpa": 7.52,
  "hobbies": ["music", "workout"],
  "weakness": "overthinking"
}
```

```python
# Printing all keys
for key in info.keys():
  print(key, end=", ")
```

```python
# Printing all values
for value in info.values():
  print(value, end=", ")
```
