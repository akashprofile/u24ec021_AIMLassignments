# Task

Write a program to count vowels in a string.

```python
vowels = {"a", "e", "i", "o", "u"}

strings = [
  "Oggy",
  "Jack",
  "Joey",
  "Marie",
  "Didi",
  "Ben Tennyson",
  "Gwen Tennyson",
  "Kewin lewin",
  "Josee, the Tiger and the Fish"
]

max_str_length = 0
for value in strings:
  length = len(value)
  if (max_str_length < length):
    max_str_length = length

max_str_length
```

```python
vowels_counts = []

for string in strings:
  chars = list(string.lower())
  vowels_count = 0
  for char in chars:
    if (char in vowels):
      vowels_count += 1
  vowels_counts.append(vowels_count)

vowels_counts
```
