## 8. Create at least 2 countplots/bar plots

```python
fig, axs = plt.subplots(1, 2, figsize=(18, 6))

data = df["Inventory_Level"].value_counts()
bars = axs[0].bar(data.index.to_list(), data.values)
axs[0].bar_label(bars)
axs[0].set_xlabel("Inventory_Level")
axs[0].set_ylabel("Count")

data = df["SupplyChain_Efficiency_Label"].value_counts()
bars = axs[1].bar(data.index.to_list(), data.values)
axs[1].bar_label(bars)
axs[1].set_xlabel("SupplyChain_Efficiency_Label")
axs[1].set_ylabel("Count")
0
```

## 10. Create at least 1 scatter plot.

```python
fig, axes = plt.subplots(figsize=(6, 4))

axes.scatter(df["Resource_Utilization"].values, df["Delivery_Efficiency"].values)
axes.set_xlabel("Resource_Utilization")
axes.set_ylabel("Delivery_Efficiency")
```

## 11. Create a correlation heatmap.

```python
numericalFeaturesName = [feature.name for feature in numericalFeatures]

corr = df.loc[:, numericalFeaturesName].corr()

fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(
  corr,
  annot=True,
  cmap="coolwarm",
  vmin=-1,
  vmax=1,
  ax=ax
)

ax.set_title("Correlation Heatmap")
```
